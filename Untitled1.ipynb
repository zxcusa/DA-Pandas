```python
import pandas as pd
data = {
    "maydon": [40, 55, 60, 75, 80, 95, 110, 120, 135, 150],
    "narxi": [320, 420, 460, 540, 600, 6580, 760, 830, 900, 980],
}
df = pd.DataFrame(data)
df.to_csv("uy_narxlari.csv",index=False)
print("Fayl yaratildi!")
```

```python
dp =pd.read_csv("uy_narxlari.csv")
print(df.head())
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = df[['maydon']]
y = df['narxi']

X_train,X_test,y_train,y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(X_train,y_train)

print("w(coef):",model.coef_)
print("b(intrecept):",model.intercept_)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error,r2_score
```

```python
mse = mean_squared_error(y_test,y_pred)
rmse = mse ** 0.5
r2=r2_score(y_test,y_pred)

print(f"MSE: {mse:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R2: {r2:3f}")
```

```python
import pandas as pd
data1 = {
    "vaqt": [2, 4, 6, 8, 10, ],
    "ball": [55, 68, 74, 89, 95, ],
}
df = pd.DataFrame(data1)
df.to_csv("ballar.csv",index=False)
print("Fayl yaratildi!")
```

```python
dp =pd.read_csv("ballar.csv")
print(df.head())
```

```python
x = df[['vaqt']]
y = df['ball']

X_train,X_test,y_train,y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)

print("w(coef):",model.coef_)
print("b(intrecept):",model.intercept_)
```

```python
y_pred = model.predict(X_test)
```

```python
mse = mean_squared_error(y_test,y_pred)
rmse = mse ** 0.5
r2=r2_score(y_test,y_pred)

print(f"MSE: {mse:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R2: {r2:3f}")
```
